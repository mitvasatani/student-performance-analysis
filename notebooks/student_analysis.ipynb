```python
df["total_score"] = (
    df["math_score"] +
    df["reading_score"] +
    df["writing_score"]
)
```

```python
df["average_score"] = df["total_score"] / 3
```

```python
import matplotlib.pyplot as plt
```

```python
subjects = ["Math", "Reading", "Writing"]

averages = [
    df["math_score"].mean(),
    df["reading_score"].mean(),
    df["writing_score"].mean()
]
```

```python
plt.bar(subjects, averages)

plt.title("Average Score by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Score")

plt.show()
```

```python
plt.scatter(
    df["study_hours"],
    df["average_score"]
)

plt.title("Study Hours vs Average Score")
plt.xlabel("Study Hours")
plt.ylabel("Average Score")

plt.show()
```

```python
plt.scatter(
    df["attendance"],
    df["average_score"]
)

plt.title("Attendance vs Average Score")
plt.xlabel("Attendance (%)")
plt.ylabel("Average Score")

plt.show()
```

```python
import seaborn as sns
```

```python
sns.regplot(
    data=df,
    x="study_hours",
    y="average_score"
)

plt.title("Study Hours vs Average Score")
plt.show()
```

```python
df["study_hours"].corr(df["average_score"])
```

```python
correlation = df[
    [
        "study_hours",
        "attendance",
        "math_score",
        "reading_score",
        "writing_score",
        "average_score"
    ]
].corr()

sns.heatmap(
    correlation,
    annot=True
)

plt.title("Correlation Between Student Performance Factors")

plt.savefig("../images/correlation_heatmap.png")
plt.show()
```
